```python
data_score = pd.read_csv('imgs_data.csv')
data_score.head()
```

```python
sel_data = sel_data[['right','day','time','long_right','lat_right']]
```

```python
data_score['filename'] = data_score['filename'].str[:-4]
```

```python
sel_data = sel_data.merge(data_score,left_on = 'right',right_on = 'filename',how = 'left')
```

```python
sel_data.head()
```

```python
sel_data.columns
```

```python
from folium.plugins import HeatMap

m = folium.Map(location = [sel_data['long_right'].mean(),sel_data['lat_right'].mean()],
               zoom_start = 15,
               tiles = 'CartoDB positron')

# 构建热力图数据[lat, lon, weight]
heat_data = []
for idx,row in sel_data.iterrows():
    if not pd.isna(row['elo_score_safety']): # 排除缺失值
        heatdata = [row['long_right'], row['lat_right'], row['elo_score_safety']]
        heat_data.append(heatdata)

# 添加热力图
HeatMap(heat_data,
       min_opacity = 0.2,
        radius = 15,
        blur = 10,
        max_zoom = 1
       ).add_to(m)

m.save('map_04.html')
m
```

```python
sel_data['hour'] = sel_data['time'].str.split(':').str[0].astype(int)

# 白天（6 <= hour < 18）
sel_data_day = sel_data[(sel_data['hour'] >= 6) & (sel_data['hour'] < 18)]
sel_data_night = sel_data[(sel_data['hour'] < 6) | (sel_data['hour'] >= 18)]
```

```python
m = folium.Map(location = [sel_data_night['long_right'].mean(),sel_data_night['lat_right'].mean()],
              zoom_start = 10,
              tiles = 'CartoDB dark_matter')

# 构建热图数据：[lat, lon, weight]
heat_data = []
for idx,row in sel_data.iterrows():
    if not pd.isna(row['elo_score_safety']): # 排除缺失值
        heatdata = [row['long_right'], row['lat_right'], row['elo_score_safety']]
        heat_data.append(heatdata)

# 添加热力图
HeatMap(heat_data,
        min_opacity=0.2,
        radius=15,
        blur=10,
        max_zoom=1).add_to(m)

m.save('map_05.html')
m
```

```python
m = folium.Map(location=[ sel_data_night['long_right'].mean(),sel_data_night['lat_right'].mean()], 
               zoom_start=15,
               tiles='CartoDB dark_matter')

# 构建热图数据：[lat, lon, weight]
heat_data = [
    [row['long_right'], row['lat_right'], row['elo_score_safety']]
    for idx, row in sel_data_night.iterrows()
    if not pd.isna(row['elo_score_safety'])  # 排除缺失值
]

# 添加热力图
HeatMap(
    data=heat_data,
    radius=15,           # 每个点的扩散范围
    blur=10,             # 模糊程度
    min_opacity=0.2,     # 最低透明度（背景点是否可见）
    max_zoom=14,         # 最大缩放级别仍显示热图
    gradient={
        0.2: 'blue',
        0.4: 'lime',
        0.6: 'yellow',
        0.8: 'orange',
        1.0: 'red'
    },
    name='Elo Score Heatmap', # 图层名称
    overlay=True, # 是否为图层叠加 一般默认为 true
    show=True # 是否初始显示该热图
).add_to(m)

m.save('map_06.html')
m
```
